```python
import os
import sys
from dotenv import load_dotenv
from openai import OpenAI

# Add the repo root to path so we can import from shared/
sys.path.append(os.path.abspath(".."))
from shared.shopbot_config import SHOPBOT_SYSTEM_PROMPT

load_dotenv("../.env")
client = OpenAI()

# gpt-4o-mini is cheap and fast
MODEL = "gpt-4o-mini"
```

```python
def ask_shopbot(user_message: str) -> str:
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": SHOPBOT_SYSTEM_PROMPT},
            {"role": "user", "content": user_message},
        ],
        # Higher temperature makes the model more "creative"
        temperature=1.0,
    )
    return response.choices[0].message.content
```

```python
gap_questions = [
    "How many points do I earn per dollar I spend, and how many points do I need to get $10 off?",
]

for q in gap_questions:
    print(f"Question: {q}")
    print(f"ShopBot:  {ask_shopbot(q)}")
    print("-" * 60)
```
